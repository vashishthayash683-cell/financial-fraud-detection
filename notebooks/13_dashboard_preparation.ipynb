Step 13.2 — Notebook ka first cell

```python
import pandas as pd
import numpy as np
import os
import joblib

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
```

Step 13.3 — Dataset load karo

```python
df = pd.read_csv("../data/raw/financial_fraud_detection_dataset.csv")

print("Dataset Shape:", df.shape)
print(df.head())
```

Step 13.4 — Target aur features

```python
target_column = "Fraudulent"

X = df.drop(columns=[target_column])
y = df[target_column]

print("X Shape:", X.shape)
print("y Shape:", y.shape)
```

Step 13.5 — IDs aur Date remove karo

```python
X = X.drop(
    columns=["Transaction_ID", "Customer_ID", "Transaction_Date"]
)

print("Features after dropping ID/date:")
print(X.columns.tolist())
print("Shape:", X.shape)
```

Step 13.6 — Train/Test split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

Step 13.7 — Numeric & categorical columns

```python
numeric_features = [
    "Transaction_Amount",
    "Is_International",
    "Previous_Transactions",
    "Average_Spend",
    "Account_Age_Days"
]

categorical_features = [
    "Merchant_Category",
    "Payment_Method",
    "Device_Type",
    "Location",
    "Suspicious_Keyword"
]
```

Step 13.8 — Preprocessor create karo

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            numeric_features
        ),
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        )
    ]
)
```

Step 13.9 — Preprocessor ko train data par fit karo

```python
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Processed training shape:", X_train_processed.shape)
print("Processed testing shape :", X_test_processed.shape)
```

Step 13.10 — Preprocessor save karo

```python
os.makedirs("../models", exist_ok=True)

preprocessor_path = "../models/preprocessor.pkl"

joblib.dump(
    preprocessor,
    preprocessor_path
)

print("Preprocessor saved successfully!")
print(preprocessor_path)
```

Step 13.11 — Sabse important verification

```python
loaded_preprocessor = joblib.load(
    "../models/preprocessor.pkl"
)

sample = X_test.iloc[[0]]

sample_processed = loaded_preprocessor.transform(sample)

print("Original sample shape:", sample.shape)
print("Processed sample shape:", sample_processed.shape)
```
