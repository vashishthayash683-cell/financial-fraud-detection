Step 11.1 — Imports

```python
import pandas as pd
import numpy as np
import joblib
import os
```

Step 11.2 — Load Final Model

```python
final_model = joblib.load(
    "../models/final_fraud_model.pkl"
)

print("Final model loaded successfully.")
print(type(final_model))
```

Step 11.3 — Load Final Threshold

```python
threshold_config = pd.read_csv(
    "../models/final_model_threshold.csv"
)

threshold_config
```

Step 11.4 — Extract Threshold

```python
final_threshold = threshold_config.loc[
    0,
    "Threshold"
]

print("Final threshold:", final_threshold)
```

Step 11.5 — Load Test Data

```python
X_test = pd.read_csv(
    "../data/processed/X_test_processed.csv"
)

y_test = pd.read_csv(
    "../data/processed/y_test.csv"
).squeeze()

print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)
```

Step 11.6 — Verify Model Input Features

```python
print("Number of model input features:", final_model.n_features_in_)
print("Number of test features:", X_test.shape[1])
```

```python

if final_model.n_features_in_ == X_test.shape[1]:
    print("Feature count verification passed.")
else:
    print("Feature count mismatch.")
```

Step 11.7 — Test Final Model

```python
test_probabilities = final_model.predict_proba(
    X_test
)[:, 1]

test_predictions = (
    test_probabilities >= final_threshold
).astype(int)

print("Predictions generated successfully.")
```

Step 11.8 — Verify Final Metrics

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

print("Final Model Performance")
print("=" * 40)

print(
    "Accuracy :",
    round(accuracy_score(y_test, test_predictions), 4)
)

print(
    "Precision:",
    round(
        precision_score(
            y_test,
            test_predictions,
            zero_division=0
        ),
        4
    )
)

print(
    "Recall   :",
    round(
        recall_score(
            y_test,
            test_predictions,
            zero_division=0
        ),
        4
    )
)

print(
    "F1 Score :",
    round(
        f1_score(
            y_test,
            test_predictions,
            zero_division=0
        ),
        4
    )
)

print(
    "ROC-AUC  :",
    round(
        roc_auc_score(
            y_test,
            test_probabilities
        ),
        4
    )
)
```

Step 11.9 — Save Model Metadata

```python
model_metadata = {
    "model_name": "Logistic Regression",
    "model_file": "final_fraud_model.pkl",
    "classification_threshold": float(final_threshold),
    "number_of_features": int(final_model.n_features_in_),
    "accuracy": 0.8850,
    "precision": 0.431655,
    "recall": 0.6250,
    "f1_score": 0.510638,
    "roc_auc": 0.799514
}

model_metadata
```

Step 11.10 — Save Metadata

```python
import json

with open(
    "../models/final_model_metadata.json",
    "w"
) as file:

    json.dump(
        model_metadata,
        file,
        indent=4
    )

print("Model metadata saved successfully.")
```

Step 11.11 — Final Verification

```python
print("FINAL MODEL PACKAGE")
print("=" * 50)

print("Model:", model_metadata["model_name"])
print("Features:", model_metadata["number_of_features"])
print("Threshold:", model_metadata["classification_threshold"])

print("\nPerformance:")
print("Accuracy :", model_metadata["accuracy"])
print("Precision:", model_metadata["precision"])
print("Recall   :", model_metadata["recall"])
print("F1 Score :", model_metadata["f1_score"])
print("ROC-AUC  :", model_metadata["roc_auc"])
```
